# ORI-C — démonstration minimale reproductible

Ce notebook ne crée aucun nouveau résultat. Il clone le dépôt public, hydrate Git LFS, installe l'environnement verrouillé puis exécute `demo_minimale.py` via le générateur de rapport HTML. Les trois résultats phares sont recalculés depuis les données versionnées.


```python
import os, pathlib, subprocess, sys
root = pathlib.Path('/content/ORI-C')
if not root.exists():
    subprocess.run(['git','clone','https://github.com/dalozedidier-dot/ORI-C.git',str(root)], check=True)
os.chdir(root)
subprocess.run(['git','lfs','pull'], check=True)
print(subprocess.run(['git','rev-parse','HEAD'], capture_output=True, text=True, check=True).stdout.strip())
```

```python
subprocess.run([sys.executable,'-m','pip','install','-r','plateforme/source_corrigee/requirements-lock.txt'], check=True)
```

```python
report = pathlib.Path('/content/oric-demo.html')
subprocess.run([sys.executable,'scripts/demo_minimale_html.py','--output',str(report)], check=True)
print(report)
```

```python
from IPython.display import HTML, display
display(HTML(report.read_text(encoding='utf-8')))
```

## Lecture

Le rapport distingue explicitement résultat de modèle, résultats empiriques rétrospectifs et résultats négatifs. Un succès de ce notebook prouve la reproductibilité de la démonstration sur le commit cloné, pas une validation générale d'ORI-C.
