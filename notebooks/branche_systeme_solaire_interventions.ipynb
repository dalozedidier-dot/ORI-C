# ORI-C — Système solaire : intervention vs bruit numérique
Recalcule le ratio certifié C-AST-01 depuis la sortie de robustesse. Il est distinct du plancher d’ensemble du CSV contrefactuel.


```python
from pathlib import Path
import json
ROOT=Path.cwd()
if not (ROOT/'plan_directeur').exists(): ROOT=ROOT.parent
p=ROOT/'plan_directeur/campagne_maximale_trois_branches/resultats/systeme_solaire_robustesse.json'
d=json.loads(p.read_text(encoding='utf-8'))
sep=d['numerical_effect_separation']
noise=sep['largest_selected_numerical_rmse']
rows=[(x['job'],x['intervention_rmse'],x['intervention_rmse']/noise) for x in sep['intervention_to_numerical_ratios']]
rows
```

```python
minimum=min(r[2] for r in rows)
print('minimum recalculé =',minimum)
assert abs(minimum-4964.415695208393) < 1e-9
```
